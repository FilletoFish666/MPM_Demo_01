### `git` 与 `GitHub` 基础

你的个人电脑应当按照 [laptop setup](https://github.com/ese-ada-lovelace-2026/laptop-setup). 中的说明完成配置。在此过程中，你应当已经配置了 `git` 用户名和邮箱。（如果还没有，请查看这些说明，并在本次课程中配置好 `git` 设置。）

网上有大量优质的 `git` 和 `GitHub` 教程。在你常用的搜索引擎中输入 "git tutorial"、"github tutorial" 或类似关键词，就能找到从基础到进阶用法的指南。现在我们来练习一些基础操作。

1. 通过 `GitHub` 创建仓库（repository）

第一步是通过你的 `GitHub` 账号创建仓库。可以将仓库理解为一个纳入版本控制的独立“项目”。打开 `https://github.com/yourusername?tab=repositories` 后，你应该会看到类似下面的页面：

<img src="figures/gh1.png">

点击 `New`，确认 `Owner` 选的是你自己，然后为仓库命名，勾选添加 `README` 的选项，最后创建仓库。

现在，这个仓库就会出现在 `GitHub` 上你的仓库列表中。

**注意：** 当然，你也可以用其他方式创建仓库。例如，先在本地编写一些代码，再将代码所在文件夹初始化为 `git` 仓库，之后推送到 `GitHub`。你可以稍后尝试，但请注意，这种方式稍微复杂一些，目前不需要采用它。

2. `clone` 仓库

要开始修改新建仓库中的内容，我们需要将它**克隆（clone）** 到本机。为此，请在终端中（Windows 用户应确认自己使用的是 WSL Ubuntu 终端），进入你希望存放克隆仓库的文件夹，然后使用命令
```
git clone https://github.com/yourusername/yourreponame.git
```
在终端中输入 `ls`（列出内容），就能看到刚刚克隆的文件夹。进入这个仓库。目前里面只有 `README.md` 文件。

3. 使用 `commit` 提交修改

如果在仓库中输入
```
git status
```
会看到类似下面的输出
```
On branch main
nothing to commit, working tree clean
```
这表示我们当前位于 `main` 分支（branch）（稍后会进一步讨论分支，目前先直接在 `main` 上操作），并且自上次 `commit` 以来没有任何修改。现在，用你喜欢的编辑器打开并修改 `README.md`（目前随便添加一些文字即可！）。保存修改并回到终端，再次输入 `git status`，观察输出有什么不同！还可以使用命令
```
git diff
```
查看每个已修改文件中的具体改动。

要让这些修改成为代码库的“正式”组成部分，需要提交（commit）它们。可以通过下面的命令完成
```bash
git commit -a -m "an appropriate commit message."
```
这会创建一个记录代码当前状态的“检查点（checkpoint）”。再次输入 `git status` 和 `git diff`，观察输出。

上面命令中的选项含义如下：

`-a`：自动将已修改或已删除的文件加入暂存区（staging area），但不包括新建文件。

`-m 'message'`：允许你在同一行中添加提交说明（commit message）。

如需了解可用选项及其用法的更多细节，请查看 `man git commit`（它会显示 `git commit` 命令的用户手册）。

要查看当前分支的提交历史，请运行
```
git log --oneline
```
每一行对应一次提交：先是用于标识提交的简短哈希值（hash），然后是提交说明。提交历史记录了代码的开发过程，因此应当提交规模较小、相对独立的改动，并在说明中写清楚改了什么、为什么改（例如“修复直方图分箱中的差一错误”，而不是“更新”）。

4. 将修改 `push` 到 `GitHub`

这些修改现在已经记录在本地，但如果查看该仓库的 `GitHub` 页面，还看不到任何变化。要让 `GitHub` 获得本地的修改，我们需要将它们 `push` 到 `remote` 仓库。使用下面的命令
```
git push
```
完成 `push` 后，回到仓库 URL 并刷新页面，就能看到这些修改了。

5. 将新文件纳入 `git` 版本控制

接下来看看如何向仓库添加新文件。在仓库中的某个位置，新建一个 `.py` 文件。目前可以写得简单一些，例如
```python
import numpy as np

a = np.sin(np.pi/2)

print(a)
```
新建文件后，再次运行 `git status`。你会看到新文件显示为 `untracked`。如果希望将未跟踪的文件纳入当前仓库的版本控制，需要先添加它，再提交，如下所示。运行
```
git add path_to/new_file.py
```
此时，在提交前运行一次 `git diff` 是个有用的练习。你会发现没有任何输出：`git diff` 比较的是文件与*暂存区*之间的差异，而 `git add` 刚刚已经把新文件放入暂存区了。要查看已暂存的内容，也就是下一次提交将包含什么，请运行 `git diff --staged`；这时就能看到新文件的内容。然后提交新文件
```
git commit -a -m "Add new_file.py"
```
如果执行 `git push`，再刷新并查看 `GitHub` 上的仓库，就会看到新文件，提交历史也已更新。

接下来，请花些时间自己仔细练习上述步骤，尝试对仓库做一些操作，有问题一定要问！

**练习：** 在仓库中新建一个 Jupyter notebook，输入几个代码单元并运行它们（例如导入 `numpy` 及其中的一些函数，再用合适的输入调用它们）。之后运行 `git status`。你观察到了什么？请注意，除了 notebook 显示为未跟踪文件之外，还生成了一些额外的元数据（metadata）文件，但我们不希望 Git 跟踪这些文件，因此小心不要添加它们！务必只添加 notebook 本身，避免用不必要的元数据文件把仓库弄得杂乱。如果我们知道这些元数据经常会生成，但又不希望它们被跟踪，能否让 Git 忽略它们？查看 [gitignore](https://git-scm.com/docs/gitignore) 文档，试试看能不能实现！

注意：如果想丢弃某个文件的所有本地修改，可以运行：
```bash
git restore <path/to/file>
```
（它会丢弃该文件中尚未通过 `git add` 暂存的修改；如果没有暂存过任何修改，文件就会恢复到上次提交时的状态。）

#### 创建分支

分支功能是版本控制软件最重要、最实用的功能之一，因此熟悉它非常重要！

简单来说，分支就是主仓库的一个独立版本（有更精确的技术定义，但目前不需要深入讨论）。例如，开发新功能时，可以创建一个分支来编写新代码，避免破坏主代码现有的功能。最终可以将它合并到主分支；如果进展不顺利，也可以删除它，主代码仍保持不变、不受影响！随着你逐渐使用分支，会发现它还有很多其他用途。

要查看当前所在的分支，可以运行
```
git branch
```
当前分支会用 \* 标记。要创建新分支，可以运行：
```
git switch -c new_branch_name
```
（现在再次运行 `git branch`，看看有什么变化！）要在已有分支间切换，可以运行
```
git switch existing_branch_name
```
例如，要从刚创建的分支返回主分支，只需运行
```
git switch main
```
（不过，修改代码之前，记得切换回你的新分支或开发分支！）

#### 将本地新建的分支推送到 `GitHub`

在本地创建新分支后，要将它推送到 `GitHub`，需要运行
```
git push -u origin new_branch_name
```
（如果想深入了解细节，记得可以查看 `man git push` 的输出！）之后，这个分支就会出现在 `GitHub` 仓库的 branches 选项卡中：

<img src="figures/gh4.png">

现在远程仓库（remote repository，此处为 `GitHub`）已经知道这个分支。如果继续向该分支推送修改，只需使用
```
git push
```
（不再需要 `-u ...` 等参数。）

#### 创建和合并拉取请求

创建和合并拉取请求是管理 `GitHub` 仓库的另一个重要方面。如果想将某个分支的修改合并到 `main` 分支，通过拉取请求来完成是良好的实践。

实现方式有很多，这里介绍一种。点击上图中的 `branches` 选项卡，你会看到类似下面的页面：

<img src="figures/gh5.png">

在想要合并的分支旁边，点击 `Create pull request` 按钮并按照提示操作。之后，在主仓库的 `Pull requests`（PR）选项卡中就能看到这个拉取请求。这个页面提供了许多实用工具，开发者可以查看代码库的修改并发表评论等。如果 PR 没有问题，打开它时，`Merge pull request` 按钮会显示为绿色：

<img src="figures/gh6.png">

**重要：** 按钮显示为绿色，并不意味着 PR 已经可以合并（它只表示 GH 知道如何在不产生冲突的情况下完成合并）。用户应仔细检查 PR 引入的修改，确保一切符合预期。

点击 `Merge pull request` 按钮后，修改就会进入 `main` 分支。之后，记得在本地切换到主分支，并运行 `git pull` 或 `git pull --ff-only`，更新本地的主分支。

**注意** `--ff-only` 表示：只有本地分支可以通过快进（fast-forward）直接与远程分支同步时，才进行更新。如果两个分支已经分叉（各自有对方没有的提交），`git` 会拒绝更新，而不是创建合并提交。如果你的远程仓库因某种原因可能不同步，这个选项会更安全。

#### 删除已合并的分支

拉取请求合并后，就不再需要它的分支了。在拉取请求页面上，`GitHub` 提供 `Delete branch` 按钮，用于删除 `GitHub` 上的分支。要删除本地副本，先切换回 `main` 并拉取更新，然后运行
```
git branch -d new_branch_name
```
`-d` 只会删除修改已经合并的分支，因此不会意外丢弃尚未合并的工作（`-D` 则会强制删除）。

#### 选项与进一步学习

我们只使用了每个命令提供的一小部分选项。上面有两项值得解释：在 `git push -u origin new_branch_name` 中，`origin` 是 `git` 为克隆来源仓库指定的名称（此处为你在 `GitHub` 上的仓库）；`-u`（`--set-upstream` 的简写）将本地分支关联到 `GitHub` 上的对应分支，之后直接运行 `git push` 或 `git pull`，就知道应当连接到哪里了。

现在还值得了解以下几项：

- `git status -s`（`--short` 的简写）：`git status` 的简洁版本，每个文件占一行。
- `git log --oneline --graph --all`：以图的形式展示所有分支的历史，便于观察分支在哪里分叉和合并。
- `git restore --staged <file>`：取消文件的暂存（撤销 `git add`），但不丢失对文件的修改。
- `git switch -`：切换回之前所在的分支。

每个命令都有自己的选项说明，因此通常不需要记住所有选项：

- `git <command> -h` 会输出命令选项的简要说明，例如 `git commit -h`；
- `git help <command>`（或 `man git <command>`）会打开完整的手册页（按 `q` 退出，按 `/` 搜索）；
- 在线资料包括[参考手册](https://git-scm.com/docs) 和免费书籍 [Pro Git](https://git-scm.com/book/en/v2).。

这个习惯同样适用于本讲的其他工具：例如，`pip install --help` 会解释你很快会用到的 `-e` 和 `-r` 选项，而 `conda env create --help` 会解释 `-f`。


## 虚拟环境

### 为什么需要虚拟环境？

现在你已经对 `Python` 的包管理器（package manager）`pip` 有了一些了解。考虑下面两个假设的包及其依赖要求：

- 包 **A** 有以下依赖：
    - 包 a，版本 >= 1.0
    - 包 b，版本 <=1.2
    - 包 c，版本 >= 2.2
    - 包 d，版本 >= 5.0
- 包 **B** 需要：
    - 包 a，版本 >= 1.0
    - 包 b，版本 >= 1.3
    - 包 e，版本 1.0
    - 包 f，版本 >= 7.0。

查看上面的要求，可以发现依赖 `package b` 存在冲突。每次使用 **A** 或 **B** 时，都通过 `pip` 切换包 `b` 的两个版本，显然不是好办法。而在实际的大型开发项目中，这类依赖冲突可能涉及几个，甚至几十个（！）包。更好的方案是同时安装软件的两个版本，并在使用 **A** 或 **B** 时，能够方便地切换到适合的环境。这正是 _虚拟环境_ 发挥作用的地方。   

虚拟环境是一种工具，通过为不同项目创建相互隔离的 Python 虚拟环境，将各项目需要的依赖分开管理。这是大多数 Python 开发者使用的最重要工具之一。

<div class="alert alert-success">

_**顺便说明**_（目前不需要深入了解），虚拟环境在激活时会修改各种 `environment variables`，即环境变量（environment variable），从而处理同一个包的不同版本等情况。 

以下内容来自 Wikipedia（https://en.wikipedia.org/wiki/Environment_variable):

_环境变量是一个名称可动态指定的值，它能够影响计算机上正在运行的进程的行为。环境变量是进程运行环境的一部分。例如，运行中的进程可以查询 TEMP 环境变量，找到存放临时文件的合适位置；也可以查询 HOME 或 USERPROFILE 变量，找到运行该进程的用户所拥有的目录结构。_

例如，一个重要的环境变量是 **PATH**。要查看它的值，可以在终端中输入
```bash
echo $PATH
```
这个变量是一组目录路径。当用户输入命令而没有提供完整路径时，系统会检查这个列表，看其中是否有能够找到该命令的路径。

总之，环境变量的值决定了环境某些方面的行为，例如默认调用或访问哪些可执行文件和库，或者执行某些命令时使用哪些选项。

</div>

### 何时、何处应该使用虚拟环境？

默认情况下，系统上的每个项目都会使用相同的目录来存放和查找 site packages（第三方库）。为什么这很重要？在上面的两个项目示例中，你有包 `b` 的两个版本。对 Python 来说，这确实是个问题，因为它无法区分 "site-packages" 目录中的不同版本。于是 v1.2 和 v1.3 会以相同的名称存放在同一个目录中。这时就需要虚拟环境了。解决方法是创建两个独立的虚拟环境，每个项目一个。好处是，环境数量没有限制，因为它们只是包含几个脚本的目录。

除了上述情况，我们还可能出于以下原因使用虚拟环境
- 有时不同包名称相同，但功能不同，会造成命名空间（namespace）冲突。
- 有时需要一个干净的环境来测试包的依赖，以便编写 `requirements.txt` 文件（稍后会进一步讨论这类文件）。

总而言之：**不要使用系统 Python 进行开发。始终在虚拟环境中工作**。

### venv
 
Python 提供了名为 `venv` 的[内置库](https://docs.python.org/3/library/venv.html)，可以用来创建所谓的“虚拟环境”。在虚拟环境中，只能使用你明确选择加入的包和工具，而且只包含你指定的版本。这样就能方便、快捷地获得一个“干净”的系统，用于测试或运行彼此不兼容的软件。

要创建一个新的 `venv` 环境，可以运行如下命令
```bash
python -m venv foo
```
或者，在没有 `python` 命令的系统上（例如不在 conda 环境中的 Ubuntu/WSL），运行
```bash
python3 -m venv foo
```

这会创建一个新目录 `foo/`，其中包含该虚拟环境相关的文件。要加载这个环境，请运行
```bash
source foo/bin/activate
```

要退出环境，请运行
```bash
deactivate
```


##### 使用 `venv` 创建 `requirements.txt` 文件

切换到 `venv` 环境，是为新项目整理一份简洁的依赖包列表的方法之一。你可以启动一个空白环境，然后尝试构建并运行代码。如果缺少依赖，程序应该会失败并产生 `ModuleNotFoundError`（一种 `ImportError`），类似于

```bash
ModuleNotFoundError: No module named 'numpy'
```

理想情况下，你能识别缺少的依赖（此处为 `numpy`），并通过运行类似 `pip install numpy` 的命令来解决。根据需要重复这些步骤，解决其他依赖缺失后，就可以生成适用于 `requirements.txt` 的 Python 环境依赖列表（使用 `pip freeze` 命令，它也会列出当前安装的版本）。

### 其他一些工具（在讨论 Anaconda 之前）

- [**virtualenv**](https://virtualenv.pypa.io/en/stable/)

这是一个常用工具，可以为 Python 库创建隔离的 Python 环境。**virtualenv** 会在目录（例如 `env/`）中安装一些文件，然后修改 `PATH` 环境变量，将自定义的 bin 目录（例如 `env/bin/`）放在最前面。这个目录中会放置 Python 二进制文件的副本或链接，而 Python 会优先在相对于自身路径的环境目录中查找库。它不属于 Python 标准库，但获得了 PyPA（Python Packaging Authority）的官方认可。激活后，可以使用 `pip` 在虚拟环境中安装包。

- [**uv**](https://docs.astral.sh/uv/)

这是一个较新的、速度非常快的工具，整合了 `pip`、`venv` 以及其他一些工具的功能。你会越来越多地遇到使用它的项目，例如使用 `uv venv` 创建环境，使用 `uv pip install numpy` 在环境中安装包，或使用 `uv sync` 创建与项目的 `pyproject.toml` 和锁文件（lock file，`uv.lock`）一致的环境。`uv` 也能安装 Python 本身，但与 `pip` 一样，它只安装 Python 包；本模块使用的 `conda` 还可以提供非 Python 依赖（C/Fortran 库、编译器等）。

- [**docker**](https://www.docker.com/)

virtualenv 只封装 Python 依赖，而 Docker 容器封装整个操作系统（operating system，OS）。使用 Python virtualenv 时，可以方便地切换 Python 版本和依赖，但宿主操作系统仍然不变。使用 Docker 镜像（image）时，则可以更换整个操作系统，在 Ubuntu、Debian、Alpine，甚至 Windows Server Core 上安装和运行 Python。网上有各种操作系统与 Python 版本组合的 Docker 镜像，可以直接拉取，在任何安装了 Docker 的系统上使用。

Docker 这类工具非常适合测试软件包以及跨操作系统、跨硬件的兼容性。对于软件开发，`conda` 这类工具通常更方便。

## 2026 年的环境文件（Python 3.14）：`pyproject.toml`、`environment.yml`，以及可能需要的 `requirements.txt`

在上面的练习中，依赖最终被手动列在 `environment.yml`、`requirements.txt` 和 `pyproject.toml`（其中一部分）中。这有助于理解各文件的作用，但也意味着要维护三份同步的列表。对于你自己的项目，我们建议采用以下方式。

### 唯一权威来源（single source of truth）
- **将 `pyproject.toml` 作为唯一权威来源**，记录项目的运行时依赖（runtime dependency）和可选扩展依赖（extras）。
- 避免在多个文件中重复维护依赖列表。

### 何时使用 `environment.yml`
使用 **`environment.yml`** 创建 **conda** 环境，固定 Python 和几个大型科学计算库的版本（version pinning）。然后，在该环境**内部**，根据 `pyproject.toml` 安装你的包及其依赖。

**推荐的 `environment.yml`：**
```bash
    name: mpm-lecture
    channels:
      - conda-forge
    dependencies:
      - python=3.14
      - pip
      - numpy>=2.1
      - pandas>=2.2.3
      - matplotlib>=3.10
      # - scipy>=1.14.1   # include only if actually used
      - pip:
          - -e .[dev]    # installs your package + extras declared in pyproject.toml
```

`channels` 指定 `conda` 从哪里下载包。`conda-forge` 是社区维护的频道，提供几乎所有科学计算包的构建版本。本模块自己的 `environment.yml` 还列出了 `nodefaults`，用于阻止 `conda` 回退到 Anaconda 的 `defaults` 频道，从而让所有包都来自 `conda-forge`。

**更新环境：**
```bash
    conda env remove -n mpm-lecture -y
    conda env create -f environment.yml
    conda activate mpm-lecture
```
> `conda env update -f environment.yml` 也可以使用，但对学生而言，其结果不那么容易预测。

### 还需要 `requirements.txt` 吗？

- 通常 **不需要**。对于 pip/venv 用户：
    `pip install -e .[dev]`
这会直接从 `pyproject.toml` 读取依赖。

- 如果需要固定版本的列表（用于 CI 或工作坊），应当**生成**它，而不是手动维护：
    `pip-compile -o requirements.txt pyproject.toml`（来自 `pip-tools`；`uv pip compile` 也能完成同样的任务）
也可以将固定版本写入**约束（constraints）** 文件，然后通过以下命令安装：
    `pip install -c constraints.txt -e .[dev]`

### 可选包与扩展依赖（优先使用 extras，而不是多个依赖文件）

在 `pyproject.toml` 中：
```bash
    [project.optional-dependencies]
    dev = ["pytest>=8", "ruff>=0.5"]
    viz = ["scikit-image>=0.24"]
```
用法：
- pip/venv：`pip install -e .[dev,viz]`
- conda（`environment.yml`）：将 `-e .[dev,viz]` 放在 `pip:` 块下。

### 为什么采用这种配置？
- **唯一权威来源**（`pyproject.toml`）→ 减少同步错误。
- **Conda** 处理 Python 和系统库；**pip** 安装你的包及其声明的依赖。
- **扩展依赖（Extras）** 让可选组件明确可见，并便于安装。
